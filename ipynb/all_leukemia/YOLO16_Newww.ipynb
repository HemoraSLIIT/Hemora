```python
from ultralytics import YOLO
```

```python
DATASET_PATH = "/content/drive/MyDrive/Research/lymphoblasts-detection-15/data.yaml"

OUTPUT_DIR = "/content/drive/MyDrive/Research/YOLO_Location_16"

model = YOLO("yolov8s.pt")

model.train(
    data=DATASET_PATH,
    epochs=120,
    imgsz=640,
    batch=8,
    patience=30,
    device=0,

    project=OUTPUT_DIR,      # custom save location
    name="wbc_detector",  # folder name
    exist_ok=True
)
```
